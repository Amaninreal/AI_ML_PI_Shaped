```python
# loading the data
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")

df.head()
```

```python
# checking with pure functions
def dataset_info(dataframe: pd.DataFrame) -> dict:
    return {
        "rows": dataframe.shape[0],
        "columns": dataframe.shape[1],
        "column_names": list(dataframe.columns)
    }
dataset_info(df)
```

```python
# adding an invalid or bad data
df.loc[0, "species"] = ""
```

```python
def remove_invalid_rows(dataframe: pd.DataFrame) -> pd.DataFrame:
    valid_indices = filter(lambda idx: dataframe.loc[idx, "species"].strip() != "", dataframe.index)
    cleaned_df = dataframe.loc[list(valid_indices)].reset_index(drop=True)
    return cleaned_df

# Cleaned version of the dataset
clean_df = remove_invalid_rows(df)
clean_df.head()
```

```python
def normalize_species(dataframe: pd.DataFrame) -> pd.DataFrame:
    df_copy = dataframe.copy()
    df_copy["species"] = df_copy["species"].map(lambda x: x.strip().title())
    return df_copy

normalized_df = normalize_species(clean_df)
normalized_df.head()
```

```python
# functional 
length_list_fp = list(
    map(lambda x: x * 2, filter(lambda x: x > 5.0, df["sepal_length"]))
)

length_list_fp[:5]
```

```python
# counting of each species 
from collections import Counter

species_counts = dict(Counter(normalized_df["species"].map(lambda x: x)))
species_counts
```

```python
# now using group by average
grouped_avg = normalized_df.groupby("species")["sepal_length"].mean().to_dict()
grouped_avg
```
