```python
df['Embarked'].unique()
```

```python
# filling 'Age' with median
df['Age'].fillna(df['Age'].median(), inplace=True)

# filling 'embarked' with median of the data
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# dropping the cabin data because it has lot of missing values
df.drop(columns=['Cabin'], inplace=True)
```

```python
df.head(30)
```

```python
#checking null sum data after filling
df.isnull().sum()
```

```python
# extracting title from names 
df['Title'] = df['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
```

```python
df.drop(columns=['PassengerId', 'Name', 'Ticket'], inplace=True)
```

```python
df.info()
```

```python
# Convert 'Sex' and 'Embarked' to numeric using map
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
df['Embarked'] = df['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
```

```python
df.head(10)
```

```python
df.groupby('Sex')['Survived'].mean() # grouping the reat of gender & avg 
```

```python
df.groupby('Pclass')['Survived'].mean() # survival range by passenger
```

```python
# survival range by age of buckets
df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 12, 18, 35, 60, 100], labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'])
df.groupby('AgeGroup')['Survived'].mean()
```

```python
# survival range by gender
sns.barplot(x='Sex', y='Survived', data=df)
plt.title("Survival Rate by Gender")
plt.show()
```

```python
# survival range by pclass

sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival Rate by Passenger Class")
plt.show()
```

```python
# age distribution of survivors & non-survivors
sns.histplot(data=df, x='Age', hue='Survived', multiple='stack', bins=20)
plt.title("Age Distribution of Survivors vs Non-Survivors")
plt.show()
```

```python
df.head(40)
```

```python
rare_titles = ['Lady', 'Countess', 'Capt', 'Col', 'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona', 'Master']
df['Title'] = df['Title'].replace(rare_titles, 'Rare')
```

```python
df.head(40)
```

```python
# calculating survival rate by title
title_survival = df.groupby('Title')['Survived'].mean().reset_index()

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=title_survival, hue='Title', y='Survived', palette='Set2')
plt.title('Survival Rate by Title')
plt.ylabel('Survival Rate')
plt.xlabel('Passenger Title')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# count of passengers by Title and Survival
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Title', hue='Survived', palette='pastel')
plt.title('Passenger Count by Title and Survival')
plt.xlabel('Title')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.tight_layout()
plt.show()
```

```python
rare_survivors = df[df['Title'] == 'Rare']['Survived'].value_counts()

# adding lebels
labels = ['Did Not Survive', 'Survived']
colors = ['#ff9999','#66b3ff']

plt.figure(figsize=(6, 6))
plt.pie(rare_survivors, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors)
plt.title('Survival Distribution for Rare Titles')
plt.tight_layout()
plt.show()
```
